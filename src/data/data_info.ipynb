```python
import numpy as np
file = 'taobao_small.txt'
with open(file, 'r') as f:
    lines = f.readlines()

print("Number of lines in the file:", len(lines))

users = set()
items = set()

lengths = []
for line in lines:
    if line.strip():  # Check if the line is not empty
        user = line.split()[0]  # Assuming the first word is the username
        item = line.split()[1:]
        users.add(user)
        items.update(item)
        lengths.append(len(item))
print("Number of unique users:", len(users))
print("Number of unique items:", len(items))
print('lengths of items:', np.array(lengths).mean())
```
